```python
import pandas as pd

df = pd.read_csv("../data/raw/dataset.csv")
```

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")
```

```python
df["TotalCharges"].isnull().sum()
```

```python
df[df["TotalCharges"].isnull()]
```

```python
df = df.dropna(subset=["TotalCharges"])
```

```python
print(df.shape)

print(df.isnull().sum())
```

Removing customer id

```python
df.drop("customerID", axis=1, inplace=True)
```

ENCODING

```python
df["Churn"] = df["Churn"].map({
    "No": 0,
    "Yes": 1
})
```

```python
df["Churn"].value_counts()
```

```python
df.info()
```

```python
df.to_csv("../data/processed/cleaned_dataset.csv", index=False)
```

## Data Preprocessing Summary

- Converted `TotalCharges` to numeric.
- Identified 11 missing values caused by new customers with zero tenure.
- Removed those rows because they represented less than 0.2% of the dataset.
- Dropped the `customerID` column as it carries no predictive information.
- Encoded the target variable (`Churn`) into binary values (0 = No, 1 = Yes).
- Saved the cleaned dataset for subsequent modeling.
